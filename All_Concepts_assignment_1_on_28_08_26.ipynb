```
LIST []- It allows duplicate values
       - It is mutable values
       - It is Ordered
       - It can be add,Remove,replace
```

````python
name = ['Deepan','diana','ram','ravi','nisha','sibiraj','yogi',3,3.35,True,{3:4},'ravi']

print(name)
print(len(name))
print(name[4])
print(type(name))

# ADD
name.append(1)
print(name)

# REMOVE
name.remove(name[5])
print(name)

# REPLACE
name[3] = 'gopal'
print(name)

````



```
Tuple ()- It is order
        - It is unmutable
        - it can't be add,remove & replace
        - It allow the duplicate
```



````python
tup = ('Deepan','diana','ram','ravi','nisha','sibiraj','yogi',3,3.35,True,{3:4},'ravi')

print(tup)
print(len(tup))
print(tup[2])
print(type(tup))

# It can display the output using index values
print(tup[2:5])
print(tup[-5:-3])
````


```
SET {}  - It remove the duplicate
        - It is unorder
        - it can be add,remove & replace
```

````python
setname = {'Deepan','diana','ram','ravi','nisha','sibiraj','yogi',3,3.35,True,'ravi','Deepan','diana','ram','ravi'}

print(setname)
print(type(setname))

# ADD
setname.add('mani')
print(setname)

# REMOVE
setname.remove('yogi')
print(setname)
````

```
Dictionary - It paired as key & value
           - It enclosed with {}
           - key is fixed & value is change
```


````python
diction = { 'Name' : 'Deepan', 'age' : 28, 'location' : 'chennai', 'couse' : 'GEN AI'}

print(diction)
print(type(diction))
print(len(diction))
print(diction['age'])
print(diction.get('Name'))


# Acess the value & change
diction['location'] = 'banglore'
print(diction)


````

````python
# Using string formatting

all_info = {'Name' : 'Deepan',
            'age' : 28,
            'location' : 'chennai',
            'course' : 'GEN AI'}

print('I am %(Name)s ,My Age is %(age)d Located in %(location)s Study %(course)s Course' % all_info)

print('I am {} ,My Age is {} Located in {} Study {} Course'.format(all_info['Name'],all_info['age'],all_info['location'],all_info['course']))

print('I am {Name} ,My Age is {age} Located in {location} Study {course} Course'.format(**all_info))

print(f"I am {all_info['Name']}, My Age is {all_info['age']} Located in {all_info['location']} Study {all_info['course']} Course")
````
